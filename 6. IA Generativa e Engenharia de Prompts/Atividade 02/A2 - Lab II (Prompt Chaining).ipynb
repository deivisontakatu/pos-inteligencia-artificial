```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device_map="auto"
)
```

```python
def ask(prompt):
  response = pipe(
      [{"role": "user", "content": prompt}],
      max_new_tokens=400,
      do_sample=False
  )

  return response[0]["generated_text"][-1]["content"]
```

```python
# Prompt 1
etapa1 = ask('Liste os principais tópicos sobre Machine Learning.')

print(etapa1)
```

```python
# Prompt 2
etapa2 = ask(f'Com base nos tópicos abaixo, gere um resumo. {etapa1}')

print(etapa2)
```

```python
# Prompt 3
etapa3 = ask(f"""
Transforme o resumo abaixo em um texto adequado para alunos iniciantes.

{etapa2}
""")

print(etapa3)
```

Depois de executar as três células, o modelo ficou mais inteligente?

A resposta é **não**.

O modelo continua exatamente o mesmo. O que mudou foi a estratégia de interação: em vez de exigir que ele resolvesse tudo em uma única etapa, o problema foi decomposto em uma sequência de tarefas menores e mais bem definidas. Essa é justamente a essência do Prompt Chaining e prepara o caminho para os próximos conceitos da disciplina, como Self-Ask, Tree of Thoughts e ReAct, nos quais o próprio processo de raciocínio passa a ser organizado e controlado.
