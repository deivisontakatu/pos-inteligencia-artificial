```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device_map="auto"
)
```

```python
def ask(prompt):

    response = pipe(
        [{"role": "user", "content": prompt}],
        max_new_tokens=400,
        do_sample=False
    )

    print(response[0]["generated_text"][-1]["content"])
```

```python
ask("Explique Machine Learning.")
```

```python
prompt = """
Você é professor de pós-graduação.

Contexto:
Alunos iniciando em IA.

Objetivo:
Explique Machine Learning.

Instruções:
- Linguagem simples;
- Um exemplo;
- Vantagens;
- Limitações.

Formato:

1. Definição
2. Exemplo
3. Vantagens
4. Limitações

Responda em aproximadamente 150 palavras - sempre respondendo completamente os 4 tópicos.
"""

ask(prompt)
```
