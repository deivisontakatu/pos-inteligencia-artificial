```python
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
```

```python
# 1. Carrega o LDM direto na GPU
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
).to("cuda")

# 2. Executa a difusão no espaço latente
prompt = "A glowing Isometric cube in latent space, digital art, high quality"
image = pipe(prompt=prompt, num_inference_steps=25).images[0]

# 3. Exibe a imagem
image.show()
image.save("resultado_latente.png")
```
