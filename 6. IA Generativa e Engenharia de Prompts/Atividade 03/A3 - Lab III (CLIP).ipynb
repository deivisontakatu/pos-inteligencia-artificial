```python
import torch
from PIL import Image
import requests
from transformers import CLIPProcessor, CLIPModel
```

```python
# 1. Carrega o modelo CLIP compacto da OpenAI
model_id = "openai/clip-vit-base-patch32"
model = CLIPModel.from_pretrained(model_id)
processor = CLIPProcessor.from_pretrained(model_id)

# 2. Imagem de teste (baixa uma foto pública de um gato)
url = "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=500"
image = Image.open(requests.get(url, stream=True).raw)

# 3. Lista de frases para o CLIP testar
labels = ["a photo of a cat", "a photo of a dog", "a car on the street"]

# 4. Processa a imagem e os textos juntos
inputs = processor(text=labels, images=image, return_tensors="pt", padding=True)
outputs = model(**inputs)

# 5. Calcula as probabilidades (Softmax)
logits_per_image = outputs.logits_per_image
probs = logits_per_image.softmax(dim=1).detach().numpy()[0]

# 6. Exibe o resultado
print("🔍 Probabilidades calculadas pelo CLIP:")
for label, prob in zip(labels, probs):
    print(f" -> '{label}': {prob * 100:.2f}%")
```
