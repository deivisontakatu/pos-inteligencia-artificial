```python
from transformers import pipeline
from transformers.utils import logging

logging.set_verbosity_error()

gerador = pipeline(
    "text-generation",
    model="pierreguillou/gpt2-small-portuguese"
)
```

```python
texto_inicial = "A inteligência artificial pode"

resultado = gerador(
    texto_inicial,
    max_new_tokens=30
)

print(resultado[0]["generated_text"])
```

```python
texto_inicial = "Brasil - Brasília, Argentina - Buenos Aires, França - París"

resultado = gerador(
    texto_inicial,
    max_new_tokens=30
)

print(resultado[0]["generated_text"])
```
