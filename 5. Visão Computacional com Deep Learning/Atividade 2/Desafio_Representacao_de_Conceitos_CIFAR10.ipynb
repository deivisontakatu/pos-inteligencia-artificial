# Representando conceitos de Visão Computacional com CIFAR-10

Conceito 1 - dataset de classificação de imagem

```python
import torch
import torchvision
import matplotlib.pyplot as plt

from torchvision import datasets, transforms

cifar10 = datasets.CIFAR10(
    root="./data",
    train=True,
    download=True
)

classes = cifar10.classes

print("Quantidade de imagens:", len(cifar10))
print("Classes do CIFAR-10:", classes)

indice_img1 = 1234
indice_img2 = 8765

img1, label1 = cifar10[indice_img1]
img2, label2 = cifar10[indice_img2]

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.imshow(img1)
plt.title(f"Imagem 1: {classes[label1]}")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img2)
plt.title(f"Imagem 2: {classes[label2]}")
plt.axis("off")

plt.show()
```

Conceito 2 — Imagem como tensor

```python
to_tensor = transforms.ToTensor()

tensor_img1 = to_tensor(img1)
tensor_img2 = to_tensor(img2)

print("Tensor da imagem 1")
print("Shape:", tensor_img1.shape)
print("Tipo:", tensor_img1.dtype)
print("Valor mínimo:", tensor_img1.min().item())
print("Valor máximo:", tensor_img1.max().item())

print("\nTensor da imagem 2")
print("Shape:", tensor_img2.shape)
print("Tipo:", tensor_img2.dtype)
print("Valor mínimo:", tensor_img2.min().item())
print("Valor máximo:", tensor_img2.max().item())
```

Conceito 3 — Rótulo ou classe da imagem

```python
crop_manual = img1.crop((5, 5, 25, 25))

plt.figure(figsize=(6,3))

plt.subplot(1,2,1)
plt.imshow(img1)
plt.title("Imagem original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(crop_manual)
plt.title("Fatiamento da imagem")
plt.axis("off")

plt.show()
```

Conceito 4 — Aumento por espelhamento horizontal

```python
flip_horizontal = transforms.RandomHorizontalFlip(p=1.0)

img1_flip = flip_horizontal(img1)
img2_flip = flip_horizontal(img2)

plt.figure(figsize=(8, 4))

plt.subplot(2, 2, 1)
plt.imshow(img1)
plt.title("Imagem 1 original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(img1_flip)
plt.title("Imagem 1 espelhada")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(img2)
plt.title("Imagem 2 original")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(img2_flip)
plt.title("Imagem 2 espelhada")
plt.axis("off")

plt.show()
```

Conceito 5 — Aumento por recorte aleatório

```python
recorte_aleatorio = transforms.RandomResizedCrop(
    size=(32, 32),
    scale=(0.5, 1.0),
    ratio=(0.75, 1.33)
)

img1_crop = recorte_aleatorio(img1)
img2_crop = recorte_aleatorio(img2)

plt.figure(figsize=(8, 4))

plt.subplot(2, 2, 1)
plt.imshow(img1)
plt.title("Imagem 1 original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(img1_crop)
plt.title("Imagem 1 com recorte")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(img2)
plt.title("Imagem 2 original")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(img2_crop)
plt.title("Imagem 2 com recorte")
plt.axis("off")

plt.show()
```

Conceito 6 — Conversão para escala de cinza

```python
from torchvision.transforms import Grayscale

gray_transform = Grayscale(num_output_channels=1)

img1_gray = gray_transform(to_tensor(img1))
img2_gray = gray_transform(to_tensor(img2))

plt.figure(figsize=(8,4))

plt.subplot(2,2,1)
plt.imshow(img1)
plt.title("Imagem 1 original")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(img1_gray.squeeze(), cmap="gray")
plt.title("Imagem 1 em escala de cinza")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(img2)
plt.title("Imagem 2 original")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(img2_gray.squeeze(), cmap="gray")
plt.title("Imagem 2 em escala de cinza")
plt.axis("off")

plt.show()
```
