# **Dataset Nile**

O conjunto de dados **Nile** é outro clássico absoluto no ensino de séries temporais, servindo como um excelente contraponto ao **AirPassengers**. Enquanto o **AirPassengers** é ótimo para observar a sazonalidade,enquanto que o **Nile** é perfeito observar  a  mudanças estruturais e tendência.


## **Visão Geral do Dataset Nile**

* **Conteúdo**	Vazão anual do Rio Nilo (medida em 10^8 m3) em um ponto  specífico no Sudão.

* **Período** 	De 1871 a 1970 (100 observações).

* **Frequência	Anual** (1 observação por ano).

* **Número de Observações** 	100 anos.


* **Ausência de Sazonalidade** Como os dados são anuais, não há componente sazonal (que exige periodicidade intra-anual). Isso permite que o aluno foque apenas em tendência, nível e ruído, sem a complexidade adicional da sazonalidade.

* **Mudança de Nível (Level Shift):**  Por volta de 1898, há uma queda brusca e permanente na vazão média do rio. Acredita-se que isso tenha sido causado pela construção da primeira barragem de Assuã (Aswan Low Dam) em 1898-1902. Essa mudança estrutural é o ponto central da análise deste dataset


## **Realize a decomposição da série Rio Nilo**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.seasonal import seasonal_decompose
from  matplotlib.pylab import rcParams
rcParams['figure.figsize'] =15,7
```

```python
import statsmodels.api as sm
```

```python
nile = sm.datasets.get_rdataset('Nile').data
```

```python
nile.head(10 )
```

```python
serie_nilo = pd.Series(nile['value'].values, index=nile['time'], name='Rio_Nilo')
```

```python
serie_nilo.head(10)
```

```python
# Plotar a série
plt.plot(serie_nilo)
plt.xlabel('Ano')
plt.ylabel('Vazão (10^8 m³)')
plt.title('Vazão do Rio Nilo (1871-1970)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

#**Dataset Shampoo Sales**

O conjunto de dados Shampoo Sales é um pequeno e clássico dataset de séries temporais, muito utilizado em tutoriais introdutórios de previsão com redes neurais e modelos **ARIMA.** Ele foi popularizado por Jason Brownlee em seu site Machine Learning Maste


## **Visão Geral do Dataset**


* **Conteúdo**	Vendas mensais de um shampoo ao longo de 3 anos.
* **Período**	De janeiro do ano 1 a dezembro do ano 3 (36 observações).
* **Frequência**	Mensal (12 observações por ano).
* **Número de Observações**	36 meses.

## **Características e Aplicações**

1. Tendência de Crescimento Há uma clara tendência de crescimento nas vendas ao longo dos 3 anos, com as vendas aumentando de cerca de 100-200 unidades/mês no início para mais de 600 unidades/mês no final do período.

2. Padrão Sazonal Anual A série apresenta um padrão sazonal anual com picos no final do ano (novembro/dezembro) e vales no início do ano (janeiro/fevereiro), sugerindo um efeito de sazonalidade de fim de ano.

3. Variância Crescente com a Tendência Assim como no AirPassengers, a amplitude das variações sazonais aumenta com o nível da série, sugerindo um comportamento multiplicativo. Isso é evidenciado pelos valores crescentes dos picos e vales ao longo do tempo.

## **Realize a decomposição da série Shampoo Sales**

```python
import pandas as pd

# Carregar diretamente da URL
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/shampoo.csv"
df = pd.read_csv(url)

# Visualizar as primeiras linhas
print(df.head())
```

```python
indice = pd.date_range(start='2021-01-01', periods=len(df), freq='MS')
```

```python
serie_shampoo = pd.Series(df['Sales'].values, index=indice, name='Vendas Shampoo')
```

```python
plt.plot(serie_shampoo)
plt.xlabel('Período')
plt.ylabel('Vendas')
plt.title('Vendas do Shampoo')
plt.grid(alpha=0.3)
```

# **Dataset UKgas**

O conjunto de dados UKgas é um clássico no estudo de séries temporais, frequentemente utilizado para demonstrar a análise de componentes como tendência e sazonalidade trimestral, com um padrão sazonal particularmente interessante e didático.

## **Visão Geral do Dataset UKgas**
* **Conteúdo Consumo trimestral** de gás no Reino Unido (em milhões de termas).

* **Período **Do primeiro trimestre de 1960 ao quarto trimestre de 1986 .

* **Frequênci**a Trimestral (4 observações por ano).

* **Número de Observações** 108 observações (27 anos × 4 trimestres).

* **Tendência de Crescimento** Ao longo dos 27 anos, há uma clara tendência de crescimento no consumo de gás, refletindo a expansão do uso de gás natural no Reino Unido durante as décadas de 1960, 1970 e 1980.

* **Sazonalidade Trimestral** com Padrão de "Onda" O padrão sazonal é o grande destaque deste dataset. O consumo de gás é significativamente maior no inverno (primeiro e quarto trimestres) e menor no verão (terceiro trimestre), devido à necessidade de aquecimento residencial e comercial. Isso cria um padrão em forma de "onda" ou "serra" que é facilmente identificável visualmente.

##  **Realize a decomposição da série UKgas**

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Carregar o dataset
ukgas_data = sm.datasets.get_rdataset('UKgas')
df = ukgas_data.data
```

```python
serie_ukgas = pd.Series(df['value'].values, index=df['time'], name='Vendas Shampoo')
```

```python
plt.plot(serie_ukgas, linewidth=1.5)
plt.xlabel('Ano')
plt.ylabel('Consumo de Gás (milhões de termas)')
plt.title('Consumo Trimestral de Gás no Reino Unido (1960-1986)')
plt.grid(alpha=0.3)
plt.show()
```
