# **Criando uma série em Pyhton**

## **pd.Series**
Um **pd.Series** é uma das estruturas de dados fundamentais da biblioteca Pandas em Python. Pode ser pensado como uma coluna de uma tabela, um vetor unidimensional rotulado ou um dicionário otimizado de alta performance.


**Características Principais**
* **Unidimensional:** Armazena uma sequência de dados de uma única dimensão (uma linha ou uma coluna).

* **Homogéneo:** Todos os elementos dentro de uma Series possuem o mesmo tipo de dado (int, float, string, datetime, etc.).

* **Rotulado (Indexado):** Diferente de uma lista comum do Python, cada elemento possui um rótulo associado chamado Index (índice), que facilita a busca e o alinhamento de dados.


##**pd.Index**

O pd.Index é o objeto do Pandas responsável por armazenar os rótulos (labels) dos eixos de uma pd.Series ou pd.DataFrame.

Enquanto os dados da tabela ficam armazenados nas colunas, o pd.Index é a estrutura que dá "nome" a cada linha (índice de linhas) ou a cada coluna (índice de colunas).

Tipos Comuns de Índices O Pandas possui subclasses otimizadas para diferentes tipos de dados:

* **pd.RangeIndex:** O índice numérico padrão (0, 1, 2, ...,  N−1 ). Economiza memória por ser uma sequência matemática.

* **pd.DatetimeIndex**: Especializado para datas e horas. Habilita funcionalidades de séries temporais como remarcação de frequência (resample) e seleção por intervalos de tempo.

* **pd.MultiIndex** Índice multinível (hierárquico), que permite representar dados multidimensionais numa estrutura 2D.

* **pd.CategoricalIndex**: Ideal para rótulos repetitivos de categorias com número limitado de opções.
