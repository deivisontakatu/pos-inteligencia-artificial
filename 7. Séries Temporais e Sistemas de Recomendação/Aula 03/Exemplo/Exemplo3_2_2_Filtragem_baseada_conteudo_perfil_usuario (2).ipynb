## **KNN Content-Based**

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics.pairwise import cosine_similarity
```

## **Construindo o perfil do usuário**.

Aqui, vamos generalizar a estraégia vista no exemplo anterior. Poranto, precisamos construir uma função que crie o perfil do usuário, selecionando  os itens que o usuário avaliou  e calcule a média poderada de suas features.


```python
# user_id -> contem a identificação do usuário
# interacoes -> Itens que ousuário assistiu e avaliou
# X_itens_df -> Contem todo os filmes do catalogo já processado.



def construir_perfil(user_id, interacoes, X_itens_df):
    """Perfil = média ponderada das features dos itens avaliados."""
    hist = interacoes[interacoes['user_id'] == user_id]
    if len(hist) == 0:
        return None

    idx = hist['item_id'].values   # índice dos itens
    notas = hist['nota'].values

    perfil = np.average(X_itens_df.iloc[idx].values, axis=0, weights=notas)
    return perfil.reshape(1, -1)
```

## **Recomendar com o KNN**

```python
def recomendar_content_based(user_id, interacoes, X_itens_df, filmes, k=10, top_n=5):
    """Recomenda top_n filmes para um usuário via KNN Content-Based."""

    perfil = construir_perfil(user_id, interacoes, X_itens_df)
    if perfil is None:
        return None

    # CORREÇÃO: treinar com .values para evitar warning de feature names
    knn = NearestNeighbors(n_neighbors=k, metric='cosine', algorithm='brute')
    knn.fit(X_itens_df.values)

    dist, idx = knn.kneighbors(perfil)  # perfil já é numpy array
    sims = 1 - dist[0]

    # Montar resultado (idx são posições 0..N-1 em X_itens_df/filmes)
    recs = filmes.iloc[idx[0]].copy()
    recs['similaridade'] = sims

    # Filtrar já vistos
    ja_vistos = interacoes[interacoes['user_id'] == user_id]['item_id'].tolist()
    recs = recs[~recs['item_id'].isin(ja_vistos)]

    return recs.sort_values('similaridade', ascending=False).head(top_n)
```

## **Executando as recomendações**

```python
for user in ['U001', 'U005', 'U007', 'U009']:
    print(f"\n Recomendações para {user}:")
    recs = recomendar_content_based(user, interacoes, X_itens_df, filmes, k=10, top_n=3)
    for _, row in recs.iterrows():
        print(f"   {row['titulo']:25s} ({row['genero']:8s}) — sim: {row['similaridade']:.3f}")
```
