# **Exercício LSTM - Daily Climate Time Series Data**


O "Daily Climate Time Series Data" é um conjunto de dados de séries temporais que contém registros climáticos diários da cidade de Delhi, Índia, no período de 1° de janeiro de 2013 a 31 de dezembro de 2017.



## **Características principais**
* Período: 2013–2017 (5 anos completos).

* Frequência: Diária.

* Localização: Delhi, Índia.

* Número de registros: 1.461 entradas (5 anos × 365 dias + 1 dia bissexto em 2016

## **Padrões e Sazonalidades Esperados**

* Temperatura (meantemp): Sazonalidade anual:
  * verões quentes (abril–junho: 30–40°C)
  * invernos frios (dezembro–fevereiro: 5–15°C).

* Tendência diária: variação entre dia e noite.

* Umidade (humidity):
  * Maior durante a monção (julho–setembro).
  * Menor no verão e inverno.

* Velocidade do Vento (wind_speed): Pode aumentar durante monções e tempestades.

* Pressão (meanpressure):
  * Varia com condições climáticas (ex.: baixa pressão indica possibilidade de chuva)

## **Atividade**


Item A) Construa uma rede neural LSTM  univarida, com adequada janela temporal a fim de prever  **meantemp**

item B) Construa uma rede neural LSTM  multivariada, com adequada janela temporal a fim de prever  **meantemp**, a partir das entradas  **meantemp**, 	**humidity**	**wind_speed**, 	**meanpressur**

item C) Construa uma rede neural LSTM  multivariada, com adequada janela temporal a fim de prever   **wind_speed**, 	**meanpressur**  partir das entradas  **meantemp**, 	**humidity**
