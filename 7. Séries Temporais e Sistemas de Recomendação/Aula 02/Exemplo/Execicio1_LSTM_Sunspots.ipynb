# **Exercício Sunspots**

Manchas solares são fenômenos temporários na fotosfera do Sol que aparecem como manchas mais escuras do que as áreas ao redor. Elas são regiões de temperatura superficial reduzida, causadas por concentrações de fluxo do campo magnético que inibem a convecção. As manchas solares geralmente aparecem em pares de polaridade magnética oposta. Seu número varia de acordo com o ciclo solar aproximado de 11 anos.


* **Item 1-** Relalize a analise exploratória dos dados
* **Item 2 -** Construa uma rede neural LSTM utilizando adequada janela de tempo a fim de prever as machas solares.
