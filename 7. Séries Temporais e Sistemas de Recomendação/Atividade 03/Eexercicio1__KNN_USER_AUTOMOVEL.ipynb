# **Dataset Automóveis**


Estrutura do Dataset

*  **2000 usuários**

* **10 marcas de carro**

* **Notas de 0 a 5 (podendo ser floats ou inteiros)**

* **Formato: matriz usuário × marca (User-Item Matrix)**

* **Esparsidade**

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Configurações
n_usuarios = 2000
marcas = ['Toyota', 'Volkswagen', 'Chevrolet', 'Fiat', 'Hyundai',
          'Honda', 'Ford', 'Renault', 'Nissan', 'Jeep']

# Perfis latentes (5 perfis de usuário) para gerar dados mais realistas
# Cada perfil tem preferências diferentes por marca
perfis = {
    'conservador':  [4.5, 4.0, 3.5, 2.5, 3.0, 4.2, 3.0, 2.5, 3.2, 2.8],
    'esportivo':    [3.5, 3.8, 3.0, 2.0, 3.5, 4.0, 4.5, 2.8, 4.0, 3.0],
    'economico':    [4.0, 3.5, 4.0, 4.5, 3.8, 3.5, 2.5, 4.0, 3.2, 2.0],
    'premium':      [4.8, 4.2, 3.8, 2.0, 3.5, 4.5, 3.5, 2.5, 4.0, 4.2],
    'aventureiro':  [4.0, 3.0, 3.2, 2.5, 3.5, 3.8, 3.5, 3.0, 4.2, 4.8]
}

nomes_perfis = list(perfis.keys())
pesos_perfis = [0.30, 0.15, 0.25, 0.15, 0.15]

# Cada usuário é uma mistura de perfis + ruído
dados = []
for i in range(n_usuarios):
    # Sorteia perfil base
    perfil = np.random.choice(nomes_perfis, p=pesos_perfis)
    base = np.array(perfis[perfil])

    # Adiciona variação individual (ruído gaussiano)
    ruido = np.random.normal(0, 0.6, size=len(marcas))
    notas = np.clip(base + ruido, 0, 5)

    # Simula esparsidade: ~40% dos usuários não avaliam todas as marcas
    mascara = np.random.rand(len(marcas)) > 0.4
    notas = np.where(mascara, notas, np.nan)

    dados.append(notas)

df = pd.DataFrame(dados, columns=marcas)
df.index.name = 'user_id'
df.index = [f'U{i+1:04d}' for i in range(n_usuarios)]

# Arredonda para 1 casa decimal
df = df.round(1)

print(df.shape)      # (2000, 10)
print(df.head())
```
