```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Reproduzir exatamente a base da aula
np.random.seed(42)

livros = [
    'Duna', 'Fundação', 'Neuromancer',
    'Orgulho e Preconceito', 'Como Eu Era Antes de Você',
    'Crime e Castigo', 'O Apanhador no Campo de Centeio',
    'O Pequeno Príncipe', 'A Menina que Roubava Livros', 'It: A Coisa'
]

perfis_genero = {
    'Ficção Científica': ['Duna', 'Fundação', 'Neuromancer'],
    'Romance':           ['Orgulho e Preconceito', 'Como Eu Era Antes de Você'],
    'Drama':             ['Crime e Castigo', 'O Apanhador no Campo de Centeio'],
    'Infantojuvenil':    ['O Pequeno Príncipe', 'A Menina que Roubava Livros'],
    'Terror':            ['It: A Coisa']
}

generos = list(perfis_genero.keys())
usuarios = [f'U{i:03d}' for i in range(1, 51)]

matriz_ratings = pd.DataFrame(np.nan, index=usuarios, columns=livros)

for user_id in usuarios:
    genero_fav = np.random.choice(generos)
    genero_sec = np.random.choice([g for g in generos if g != genero_fav])
    n_avaliacoes = np.random.randint(3, 9)
    livros_avaliados = np.random.choice(livros, size=n_avaliacoes, replace=False)
    for livro in livros_avaliados:
        if livro in perfis_genero[genero_fav]:
            nota = np.random.choice([4, 5], p=[0.4, 0.6])
        elif livro in perfis_genero[genero_sec]:
            nota = np.random.choice([3, 4, 5], p=[0.3, 0.4, 0.3])
        else:
            nota = np.random.choice([1, 2, 3, 4], p=[0.2, 0.3, 0.3, 0.2])
        matriz_ratings.loc[user_id, livro] = nota

```
