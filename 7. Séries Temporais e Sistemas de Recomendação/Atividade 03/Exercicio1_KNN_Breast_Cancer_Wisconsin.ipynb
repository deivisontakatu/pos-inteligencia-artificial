 # **Breast Cancer Wisconsin (Diagnostic)**

O dataset foi criado por Dr. William H. Wolberg (médico) e Olvi L. Mangasarian (cientista da computação) na Universidade de Wisconsin . Os dados originais foram coletados no University of Wisconsin Hospital entre 1989 e 1991.

* **Features:** As features foram computadas a partir de imagens digitalizadas de aspiração por agulha fina (FNA) de massas mamárias. Elas descrevem características dos núcleos celulares presentes na imagem


* **Target (Classificação)**
  * **Malignant**	Tumor maligno (câncer)	0
  * **Benign**	Tumor benigno (não-canceroso)	1



**Referência**

 Wolberg, W.H., Street, W.N., & Mangasarian, O.L. (1995). Breast Cancer Wisconsin (Diagnostic) Data Set. UCI Machine Learning Repository. https://archive.ics.uci.edu/ml/datasets/Breast+Cancer+Wisconsin+(Diagnostic)
