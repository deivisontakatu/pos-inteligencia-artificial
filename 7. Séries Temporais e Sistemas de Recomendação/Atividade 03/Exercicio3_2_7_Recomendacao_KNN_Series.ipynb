# Catalogo de série

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors

np.random.seed(7)  # semente diferente → valores diferentes

num_usuarios = 1000
series = [
    "Breaking Bad",
    "Stranger Things",
    "Game of Thrones",
    "The Wire",
    "Peaky Blinders",
    "Mad Men",
    "The Sopranos",
    "True Detective",
    "The Office",
    "Brooklyn Nine-Nine",
]

notas_possiveis = np.arange(0, 5.5, 0.5)
matriz = np.random.choice(notas_possiveis, size=(len(series), num_usuarios))
mascara = np.random.rand(len(series), num_usuarios) < 0.20
matriz[mascara] = np.nan

df_series = pd.DataFrame(
    matriz,
    index=series,
    columns=[f"Usuario_{i+1:04d}" for i in range(num_usuarios)],
)

print(df_series.iloc[:, :3])
```
