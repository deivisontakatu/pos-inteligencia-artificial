## **Importação das bibliotecas**

```python
import numpy as np
import pandas as pd
```

## **Construção da dataset de Ratings**

```python
# Definir semente para reprodutibilidade
np.random.seed(42)

# Nomes dos usuários (linhas)
usuarios = [f'Usuario_{i}' for i in range(1, 11)]

# Nomes dos filmes (colunas)
filmes = [
    'Toy Story', 'Jumanji', 'Grumpier Old Men', 'Waiting to Exhale',
    'Father of the Bride', 'Heat', 'Sabrina', 'Tom and Huck',
    'Sudden Death', 'GoldenEye'
]

# Criar matriz de avaliações (escala 1 a 5)
# Notas inteiras de 1 a 5 representando a avaliação do usuário para o filme
matriz_avaliacoes = np.random.randint(1, 6, size=(10, 10)).astype(float)

# Calcular quantos valores devem ser zero (30% de 100 = 30)
total_celulas = matriz_avaliacoes.size  # 100
qtd_zeros = int(total_celulas * 0.30)   # 30

# Sortear aleatoriamente posições para zerar (filmes não avaliados)
indices_planos = np.random.choice(total_celulas, size=qtd_zeros, replace=False)
linhas, colunas = np.unravel_index(indices_planos, matriz_avaliacoes.shape)
matriz_avaliacoes[linhas, colunas] = 0

# Criar o DataFrame
df = pd.DataFrame(matriz_avaliacoes, index=usuarios, columns=filmes)
df.index.name = 'Usuario'
```

```python
print("Dataset de reocomendação de filmes")
print(df)



```
