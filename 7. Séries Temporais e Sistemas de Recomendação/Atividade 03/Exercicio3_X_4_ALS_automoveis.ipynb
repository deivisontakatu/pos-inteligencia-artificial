```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Reproduzir resultados
np.random.seed(42)

# ============================================================
# 1. DEFINIÇÃO DAS MARCAS E CATEGORIAS (perfis)
# ============================================================
marcas = [
    # Populares / Custo-benefício
    'Fiat', 'Volkswagen', 'Chevrolet', 'Hyundai', 'Renault',
    # Premium / Luxo
    'BMW', 'Mercedes-Benz', 'Audi', 'Volvo', 'Lexus',
    # Esportivos / Alta performance
    'Porsche', 'Ferrari', 'Lamborghini', 'McLaren',
    # Elétricos / Inovação
    'Tesla', 'BYD', 'Nissan', 'Rivian',
    # SUV / Off-road
    'Jeep', 'Toyota', 'Land Rover', 'Mitsubishi'
]

perfis_categoria = {
    'Populares':      ['Fiat', 'Volkswagen', 'Chevrolet', 'Hyundai', 'Renault'],
    'Premium':        ['BMW', 'Mercedes-Benz', 'Audi', 'Volvo', 'Lexus'],
    'Esportivos':     ['Porsche', 'Ferrari', 'Lamborghini', 'McLaren'],
    'Elétricos':      ['Tesla', 'BYD', 'Nissan', 'Rivian'],
    'SUV/Off-road':   ['Jeep', 'Toyota', 'Land Rover', 'Mitsubishi']
}

categorias = list(perfis_categoria.keys())

# ============================================================
# 2. CRIAÇÃO DOS 10.000 USUÁRIOS
# ============================================================
usuarios = [f'U{i:05d}' for i in range(1, 10_001)]

matriz_ratings = pd.DataFrame(np.nan, index=usuarios, columns=marcas)

# ============================================================
# 3. GERAÇÃO DAS AVALIAÇÕES
# ============================================================
for user_id in usuarios:
    # Categoria favorita e secundária (sempre distintas)
    cat_fav = np.random.choice(categorias)
    cat_sec = np.random.choice([c for c in categorias if c != cat_fav])

    # Cada usuário avalia entre 4 e 12 marcas
    n_avaliacoes = np.random.randint(4, 13)
    marcas_avaliadas = np.random.choice(marcas, size=n_avaliacoes, replace=False)

    for marca in marcas_avaliadas:
        if marca in perfis_categoria[cat_fav]:
            # Marcas da categoria favorita → notas altas (4 ou 5)
            nota = np.random.choice([4, 5], p=[0.35, 0.65])
        elif marca in perfis_categoria[cat_sec]:
            # Marcas da categoria secundária → notas médias-altas
            nota = np.random.choice([3, 4, 5], p=[0.30, 0.45, 0.25])
        else:
            # Outras categorias → notas mais baixas
            nota = np.random.choice([1, 2, 3, 4], p=[0.25, 0.35, 0.25, 0.15])

        matriz_ratings.loc[user_id, marca] = nota

```

## **Exibindo a base de dado**s

```python
print("Shape da matriz:", matriz_ratings.shape)
print("\nPrimeiras 5 linhas:")
print(matriz_ratings.head())

print("\nPercentual de valores preenchidos:")
print(f"{(matriz_ratings.notna().sum().sum() / matriz_ratings.size) * 100:.2f}%")

print("\nMédia de avaliações por usuário:")
print(matriz_ratings.notna().sum(axis=1).mean().round(2))

print("\nNota média por marca:")
print(matriz_ratings.mean().round(2).sort_values(ascending=False))
```
