```python
import pandas as pd
import numpy as np

np.random.seed(123)

n_investidores = 50
investimentos = ['Tesouro Direto', 'CDB', 'LCI/LCA', 'Debêntures', 'Poupança']

dados = np.random.dirichlet(alpha=np.ones(len(investimentos)), size=n_investidores)
dados_pct = np.round(dados * 100, 2)

df_carteiras_alt = pd.DataFrame(
    data=dados_pct,
    index=[f'Cliente_{i+1:03d}' for i in range(n_investidores)],
    columns=investimentos
)

print(df_carteiras_alt.head())
```
