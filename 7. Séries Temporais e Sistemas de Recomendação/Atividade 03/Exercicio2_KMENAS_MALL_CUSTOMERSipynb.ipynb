# **Mall Customers Dataset**

 O **Mall Customers Dataset** é um conjunto de dados público amplamente ustilizado para segmentação de clientes (customer segmentation) com algoritmos de clustering, especialmente o K-Means. Ele simula informações de 200 clientes de um shopping center, permitindo agrupar perfis de consumo com base em renda e comportamento de gastos.



##  **Estrutura do dataset**
* **CustomerID**	Inteiro	Identificador único do cliente

* **Gender**	Texto	Gênero: Male ou Female
* **Age	(Inteiro)**	Idade do cliente (18 a 70 anos)

* **Annual Income (k$)**	Inteiro	Renda anual em milhares de dólares

* **Spending Score (1-100)**	Inteiro	Pontuação de gastos atribuída pelo shopping
