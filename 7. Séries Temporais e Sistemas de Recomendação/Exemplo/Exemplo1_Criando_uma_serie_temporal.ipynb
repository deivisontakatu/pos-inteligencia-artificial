# **Criando uma série em Pyhton**




## **pd.Series**
Um **pd.Series** é uma das estruturas de dados fundamentais da biblioteca Pandas em Python. Pode ser pensado como uma coluna de uma tabela, um vetor unidimensional rotulado ou um dicionário otimizado de alta performance.


**Características Principais**
* **Unidimensional:** Armazena uma sequência de dados de uma única dimensão (uma linha ou uma coluna).

* **Homogéneo:** Todos os elementos dentro de uma Series possuem o mesmo tipo de dado (int, float, string, datetime, etc.).

* **Rotulado (Indexado):** Diferente de uma lista comum do Python, cada elemento possui um rótulo associado chamado Index (índice), que facilita a busca e o alinhamento de dados.



## **pd.Index**

O **pd.Index** é o objeto do Pandas responsável por armazenar os rótulos (labels) dos eixos de uma pd.Series ou pd.DataFrame.

Enquanto os dados da tabela ficam armazenados nas colunas, o pd.Index é a estrutura que dá "nome" a cada linha (índice de linhas) ou a cada coluna (índice de colunas).

**Tipos Comuns de Índices** O Pandas possui subclasses otimizadas para diferentes tipos de dados:

* **pd.RangeIndex:** O índice numérico padrão (0, 1, 2, ..., $N-1$). Economiza memória por ser uma sequência matemática.

* **pd.DatetimeIndex:** Especializado para datas e horas. Habilita funcionalidades de séries temporais como remarcação de frequência (resample) e seleção por intervalos de tempo.

* **pd.MultiIndex** Índice multinível (hierárquico), que permite representar dados multidimensionais numa estrutura 2D.

* **pd.CategoricalIndex:** Ideal para rótulos repetitivos de categorias com número limitado de opções.

## **Séries com datas**


O **pd.date_range()** é uma função fundamental do Pandas utilizada para gerar uma sequência contínua ou com intervalo fixo de datas/horas. O seu resultado é um objeto pd.DatetimeIndex, sendo a principal ferramenta para construir e organizar o eixo temporal em análises de séries temporais.

**Como Funciona**
A função exige a definição de pelo menos dois destes três parâmetros principais:
*  **start:** Data inicial.
* **end:** Data final.
* **periods:** Número total de datas a gerar.

Além disso, o parâmetro **freq** define o intervalo (frequência) entre cada data gerada.

### **Exemplo 2 - Série anual**

Vamos criar uma série temporal com valores **normalmente distribuidos** com **média zero** e **variância 1**. A série se inicia no ano de **1985** e termina no ano de **2025**.

```python
  # reprodução dos resultados.
np.random.seed(44)
```

```python
# Os dados terão distribuição normal com média zero e variancia 1.

dados_v0 =np.random.normal(0,1,41)
```

```python
dados_v1 = pd.DataFrame(dados_v0,columns=['valores'])
```

```python
dados_v1.describe()
```

```python
plt.plot(dados_v0)
```

```python
indice = pd.date_range('1985', periods=len(dados_v0),freq='YE')
```

```python
indice
```

```python
serie_v1= pd.Series(dados_v1['valores'].values, index=indice)
```

```python
type(serie_v1)
```

```python
serie_v1.plot()
plt.show()
```

## **Exemplo 3 - Série mensal**

Vamos criar uma **série temporal mensal** com valores normalmente distribuidos com **média zero** e **variância 1**. A série se inicia em **janeiro de 1985** e termina no **dezembro de 1986**.

```python
np.random.seed(44)
```

```python
dados_mensal_v0 =np.random.normal(0,1,24)
```

```python
indice_mensal
```

```python
dados_mensal_v1  = pd.DataFrame(dados_mensal_v0,columns=['valores'])
```

```python
serie_mensal= pd.Series(dados_mensal_v1['valores'].values, index=indice_mensal)
```

```python
serie_mensal.plot()
plt.show()
```
