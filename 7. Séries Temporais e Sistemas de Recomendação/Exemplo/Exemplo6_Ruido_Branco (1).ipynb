# **Ruído Branco (White Noise)**

O **Ruído Branco (White Noise)** é o conceito fundamental que representa uma série temporal puramente aleatória, sem qualquer estrutura de padrão, tendência, sazonalidade ou memória temporal.

Na modelação estatística (como em modelos **ARIMA ou LSTMs**), o **ruído branco representa o estado ideal** **para os resíduos (erros de previsão): indica que o modelo conseguiu extrair toda a informação útil dos dados, restando apenas variação aleatória imprevisível.**


##  **Definição Matemátic**a

Uma série temporal $\epsilon_t$ é definida como Ruído Branco se cumprir três condições estritas:
* **Média Nula (Constante):** $\mathbb{E}[\epsilon_t] = 0$ para todo o tempo $t$.

* **Variância Constante (Homocedasticidade):**  $\text{Var}(\epsilon_t) = \sigma^2$ para todo o tempo $t$.
* **Autocovariância Nula:** $\text{Cov}(\epsilon_t, \epsilon_{t-k}) = 0$ para qualquer desfasamento $k \neq 0$.



## **Características Principais**

* **Ausência de Memória:** Saber o valor presente $\epsilon_t$ dá zero informação sobre o valor futuro $\epsilon_{t+1}$.*
*  **Estacionariedade Forte:** É o exemplo mais puro de uma série temporal estacionária.
* **Impossibilidade de Previsão:** A melhor previsão possível para qualquer instante futuro é simplesmente a média histórica da série.

* Os pontos **oscilam de forma totalmente** desordenada e imprevisível em torno do eixo zero.

* **Não há subidas/descidas contínuas** (tendências) nem padrões repetitivos (sazonalidade).

* A **dispersão dos dados** (amplitude vertical) **permanece uniform**e do início ao fim do gráfico.


```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

```python
# 1. Gerar ruído branco gaussiano
np.random.seed(42)
ruido_branco = np.random.normal(loc=0, scale=1, size=500)
```

```python
ruido_branco.mean()
```

```python
ruido_branco.std()
```

```python
# 2. Visualizar os gráficos
fig, axes = plt.subplots(3, 1, figsize=(10, 8))

# Gráfico Temporal
axes[0].plot(ruido_branco, color='black', lw=1)
axes[0].set_title('Série Temporal de Ruído Branco')
axes[0].axhline(0, color='red', linestyle='--')

# ACF
plot_acf(ruido_branco, ax=axes[1], lags=20, title='ACF - Ruído Branco')

# PACF
plot_pacf(ruido_branco, ax=axes[2], lags=20, title='PACF - Ruído Branco')

plt.tight_layout()
plt.show()
```
