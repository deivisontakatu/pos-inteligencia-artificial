# **Construíndo uma série temporal**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Definição do Índice de Tempo (3 anos diários)
np.random.seed(42)
datas = pd.date_range(start="2023-01-01", periods=365 * 3, freq="D")
n = len(datas)
t = np.arange(n)
```

```python
# 2. Construção dos Componentes Isolados
# A. Tendência Linear de Crescimento
tendencia = 0.05 * t + 100
```

```python

# B. Sazonalidade Anual (Período P = 365 dias)
sazonalidade = 10 * np.sin(2 * np.pi * t / 365)
```

```python

# C. Ruído Aleatório Gaussiano (Média 0, Desvio Padrão 2)
ruido = np.random.normal(loc=0, scale=2, size=n)
```

```python
# 3. Composição da Série Temporal Sintética (Modelo Aditivo)
serie_sintetica = tendencia + sazonalidade + ruido
```

```python
# 4. Organização em um DataFrame Pandas
df = pd.DataFrame({
    'Série_Final': serie_sintetica,
    'Tendência': tendencia,
    'Sazonalidade': sazonalidade,
    'Ruído': ruido
}, index=datas)
```

```python
# 5. Visualização dos Componentes Didáticos
fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

axes[0].plot(df.index, df['Série_Final'], color='black')
axes[0].set_title('Série Temporal Sintética Final ($Y_t$)')

axes[1].plot(df.index, df['Tendência'], color='blue')
axes[1].set_title('Componente de Tendência Isolado ($T_t$)')

axes[2].plot(df.index, df['Sazonalidade'], color='green')
axes[2].set_title('Componente de Sazonalidade Isolado ($S_t$)')

axes[3].scatter(df.index, df['Ruído'], color='red', s=10, alpha=0.5)
axes[3].axhline(0, color='black', linestyle='--')
axes[3].set_title('Componente de Ruído Aleatório ($R_t$)')

plt.tight_layout()
plt.show()
```
