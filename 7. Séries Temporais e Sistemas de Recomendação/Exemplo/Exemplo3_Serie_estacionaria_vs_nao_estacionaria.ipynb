```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statistics as stats
from statsmodels.tsa.stattools import acovf


# 1. Configuração de reprodutibilidade e datas
np.random.seed(42)
datas = pd.date_range(start="2024-01-01", periods=200, freq="D")
n = len(datas)

# 2. Geração das Séries Temporais

# A. Série Estacionária: Média constante (0), variância constante e sem tendência
ruido = np.random.normal(loc=0, scale=1.5, size=n)
serie_estacionaria = pd.Series(ruido, index=datas)

# B. Série Não-Estacionária: Média variável ao longo do tempo (Tendência de alta)
tendencia = np.linspace(start=10, stop=50, num=n)
sazonalidade = 5 * np.sin(np.linspace(0, 8 * np.pi, n))
serie_nao_estacionaria = pd.Series(tendencia + sazonalidade + ruido, index=datas)

```

# **Analíse série estácionaria**

```python
print('Analíse série estacionaria')
print(serie_estacionaria.mean())
print(serie_estacionaria.std())
autocov_values = acovf(serie_estacionaria, fft=True)

for lag, val in enumerate(autocov_values):
    print(f"Lag {lag}: {val:.4f}")
```

```python
plt.plot(serie_estacionaria.index, serie_estacionaria, color="#2b5c8f", linewidth=1.5)
plt.axhline(0, color="red", linestyle="--", alpha=0.7, label="Média Constante ($ \mu = 0$)")
plt.title("Série Temporal Estacionária (Média e Variância Constantes no Tempo)", fontsize=12, fontweight="bold")
plt.ylabel("Valor")
plt.legend(loc="upper left")
plt.grid(True, linestyle=":", alpha=0.6)
```

**Analíse série Não estácionaria**

```python
print('Analíse série estacionaria')
print(serie_nao_estacionaria.mean())
print(serie_nao_estacionaria.std())
autocov_values = acovf(serie_nao_estacionaria, fft=True)

for lag, val in enumerate(autocov_values):
    print(f"Lag {lag}: {val:.4f}")

```

```python
# Plot da Série Não-Estacionária
plt.plot(serie_nao_estacionaria.index, serie_nao_estacionaria, color="#d95f02", linewidth=1.5)
plt.plot(serie_nao_estacionaria.index, tendencia, color="black", linestyle="--", label="Tendência (Média Variável)")
plt.title("Série Temporal Não-Estacionária (Presença de Tendência e Sazonalidade)", fontsize=12, fontweight="bold")
plt.xlabel("Data")
plt.ylabel("Valor")
plt.legend(loc="upper left")
plt.grid(True, linestyle=":", alpha=0.6)
```

```python
 #3. Construção dos Gráficos
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Plot da Série Estacionária
axes[0].plot(serie_estacionaria.index, serie_estacionaria, color="#2b5c8f", linewidth=1.5)
axes[0].axhline(0, color="red", linestyle="--", alpha=0.7, label="Média Constante ($ \mu = 0$)")
axes[0].set_title("Série Temporal Estacionária (Média e Variância Constantes no Tempo)", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Valor")
axes[0].legend(loc="upper left")
axes[0].grid(True, linestyle=":", alpha=0.6)

# Plot da Série Não-Estacionária
axes[1].plot(serie_nao_estacionaria.index, serie_nao_estacionaria, color="#d95f02", linewidth=1.5)
axes[1].plot(serie_nao_estacionaria.index, tendencia, color="black", linestyle="--", label="Tendência (Média Variável)")
axes[1].set_title("Série Temporal Não-Estacionária (Presença de Tendência e Sazonalidade)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Data")
axes[1].set_ylabel("Valor")
axes[1].legend(loc="upper left")
axes[1].grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```
