# Chatbot 1 - TF-IDF + Naive Bayes (Imobiliária)


```python

# bibliotecas
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# frases de treinamento
perguntas = [
    "Quero alugar um apartamento",
    "Tem imóveis para aluguel",
    "Quero financiar uma casa",
    "Como funciona o financiamento",
    "Posso agendar uma visita",
    "Quero visitar o imóvel",
    "Vocês trabalham com venda",
    "Quero comprar um imóvel",
    "Qual o valor do condomínio",
    "O condomínio está incluso"
]

# intenções
intencoes = [
    "aluguel",
    "aluguel",
    "financiamento",
    "financiamento",
    "visita",
    "visita",
    "compra",
    "compra",
    "condominio",
    "condominio"
]

# respostas
respostas = {
    "aluguel": "Temos diversas opções de imóveis disponíveis para aluguel.",
    "financiamento": "O financiamento pode ser realizado com bancos parceiros.",
    "visita": "Você pode agendar uma visita diretamente com um corretor.",
    "compra": "Temos imóveis residenciais e comerciais disponíveis para compra.",
    "condominio": "O valor do condomínio varia conforme o imóvel."
}

# pipeline do modelo
modelo_imobiliaria = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("naive_bayes", MultinomialNB())
])

# treinamento
modelo_imobiliaria.fit(perguntas, intencoes)

# função chatbot
def chatbot_imobiliaria(texto):
    previsao = modelo_imobiliaria.predict([texto])[0]
    return previsao, respostas[previsao]
```

```python

# testes do chatbot imobiliária

testes = [
    "Quero comprar uma casa",
    "Como agendar visita",
    "Tem apartamento para aluguel"
]

for pergunta in testes:
    intencao, resposta = chatbot_imobiliaria(pergunta)
    print("Pergunta:", pergunta)
    print("Intenção:", intencao)
    print("Resposta:", resposta)
    print("-" * 50)
```

```python

# loop conversacional

while True:
    usuario = input("Você: ")

    if usuario.lower() == "sair":
        print("Chatbot encerrado.")
        break

    intencao, resposta = chatbot_imobiliaria(usuario)

    print("Intenção identificada:", intencao)
    print("Chatbot:", resposta)
```


# Chatbot 2 - Transformers + Embeddings (Plataforma de Cursos Online)

```python

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# carregamento do modelo
modelo_transformer = SentenceTransformer('all-MiniLM-L6-v2')

# base de conhecimento
dados = {
    "certificado": {
        "frases": [
            "Como emitir certificado",
            "Onde baixar meu certificado"
        ],
        "resposta": "Os certificados ficam disponíveis na área do aluno."
    },

    "matricula": {
        "frases": [
            "Como faço matrícula",
            "Quero me inscrever em um curso"
        ],
        "resposta": "Você pode realizar a matrícula diretamente pela plataforma."
    },

    "pagamento": {
        "frases": [
            "Quais formas de pagamento",
            "Posso pagar no cartão"
        ],
        "resposta": "Aceitamos cartão, PIX e boleto bancário."
    },

    "suporte": {
        "frases": [
            "Preciso de ajuda técnica",
            "Meu vídeo não abre"
        ],
        "resposta": "Nosso suporte técnico funciona 24 horas."
    }
}

# preparação dos embeddings
frases_base = []
rotulos = []

for categoria, info in dados.items():
    for frase in info["frases"]:
        frases_base.append(frase)
        rotulos.append(categoria)

embeddings_base = modelo_transformer.encode(frases_base)

# função do chatbot
def chatbot_cursos(pergunta):
    embedding_usuario = modelo_transformer.encode([pergunta])

    similaridade = cosine_similarity(
        embedding_usuario,
        embeddings_base
    )

    indice = np.argmax(similaridade)
    categoria = rotulos[indice]

    resposta = dados[categoria]["resposta"]

    return categoria, resposta
```

```python

# testes do chatbot cursos online

perguntas_teste = [
    "Como consigo meu certificado?",
    "Quero pagar com PIX",
    "Meu curso não abre"
]

for pergunta in perguntas_teste:
    categoria, resposta = chatbot_cursos(pergunta)

    print("Pergunta:", pergunta)
    print("Intenção:", categoria)
    print("Resposta:", resposta)
    print("-" * 50)
```

```python

# loop conversacional do chatbot

while True:
    usuario = input("Você: ")

    if usuario.lower() == "sair":
        print("Encerrando atendimento.")
        break

    categoria, resposta = chatbot_cursos(usuario)

    print("Intenção identificada:", categoria)
    print("Chatbot:", resposta)
```


# Comparação Final

## Qual modelo apresentou melhores resultados?
O modelo baseado em Transformers apresentou respostas mais precisas.

## Qual modelo apresentou maior facilidade de implementação?
O modelo TF-IDF + Naive Bayes foi mais simples de desenvolver.

## Qual modelo apresentou melhor compreensão semântica?
O modelo com embeddings apresentou melhor entendimento de frases diferentes.

## Diferenças observadas entre os paradigmas
O modelo clássico depende bastante das palavras usadas no treinamento.
Já o modelo com Transformers consegue compreender melhor o contexto semântico.
